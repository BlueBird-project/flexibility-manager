- chronos-forecasting
- pydantic

```python
import requests

url = "http://127.0.0.1:8000/predict/"

payload = {
    "ds": [
        "2026-01-01 00:00:00",
        "2026-01-01 01:00:00",
        "2026-01-01 02:00:00"
    ], # Timestamp Variables
    "unique_id": ["1", "1", "1"], # ID Variables
    "y": [0, 2, 3],
    "emissions": False, # False for the meantime, CodeCarbon required.
    "freq": "H", # Frequency of the data (H for hourly, D for daily, etc.)
    "forecast_horizon": 3 # Number of days
}

headers = {
    "accept": "application/json",
    "Content-Type": "application/json"
}

response = requests.post(url, json=payload, headers=headers)

print("Status:", response.status_code)
print("Response JSON:")
print(response.json())
```

```python
response_json = response.json()
```
